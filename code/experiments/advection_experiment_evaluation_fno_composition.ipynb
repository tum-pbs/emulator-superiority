```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from advection_experiment import get_all_data
import apebench
```

```python
architectures = [
    "FNO;1;50;4;gelu",
    "FNO;2;40;4;gelu",
    "FNO;3;34;4;gelu",
    "FNO;4;30;4;gelu",
    "FNO;8;22;4;gelu",
    "FNO;12;18;4;gelu",
    "FNO;16;15;4;gelu",
    "FNO;24;13;4;gelu",
    "FNO;32;11;4;gelu",
    "FNO;48;9;4;gelu",
]
advection_scenario = apebench.scenarios.difficulty.Advection()
for arch in architectures:
    print(arch, advection_scenario.get_parameter_count(arch))
```

```python
configs = [
    {
        "network": network,
        "gamma_1": gamma_1,
        "train_simulator": train_simulator,
        "train_ic_distribution": "ind;1;True",
        "test_ic_distribution": test_ic_distribution,
        "seed_init": seed_init,
        "seed_train": seed_train,
        "seed_test": seed_test,
        "seed_shuffle": seed_shuffle,
    }
    for network in architectures
    for (gamma_1, train_simulator, test_ic_distribution) in [
        # (-0.9, "fd;0.0;1", "ind;1;True"),
        (-3.0, "fd;1.0;1", "ind;1;True"),
        # (-3.0, "fd;1.0;1", "fourier;5;True;True"),
    ]
    for seed_init in range(0, 6)
    for seed_train in range(10, 12)
    for seed_test in range(20, 22)
    for seed_shuffle in range(30, 31)
]
```

```python
len(configs)
```

```python
metrics, losses = get_all_data(configs)
```

```python
metrics = pd.concat(metrics).reset_index(drop=True)
losses = pd.concat(losses).reset_index(drop=True)
```

```python
metrics["setting"] = (
    metrics["train_simulator"]
    + "--"
    + metrics["gamma_1"].astype(str)
    + "--"
    + metrics["test_ic_distribution"]
)
losses["setting"] = (
    losses["train_simulator"]
    + "--"
    + losses["gamma_1"].astype(str)
    + "--"
    + losses["test_ic_distribution"]
)
```

```python
losses["active_modes"] = losses["network"].str.split(";").str[1].astype(int)
metrics["active_modes"] = metrics["network"].str.split(";").str[1].astype(int)
```

```python
facet = sns.relplot(
    data=losses,
    x="update_step",
    y="loss",
    col="setting",
    kind="line",
    hue="active_modes",
    estimator="median",
    errorbar=("pi", 50),
)

for ax in facet.axes.flat:
    ax.set_yscale("log")
```

```python
facet = sns.relplot(
    data=metrics.query("time_step <= 100"),
    x="time_step",
    y="error",
    hue="active_modes",
    col="setting",
    kind="line",
    estimator="median",
    errorbar=("pi", 50),
)

for ax in facet.axes.flat:
    ax.set_ylim(-0.1, 1.1)
```

```python
facet = sns.relplot(
    data=metrics.query("time_step <= 50"),
    x="time_step",
    y="superiority",
    hue="network",
    col="setting",
    kind="line",
    estimator="median",
    errorbar=("pi", 50),
    palette=["#377eb8", "#4daf4a", "#e41a1c", "#ff7f00", "#984ea3", "#a65628"],
    height=4,
)

for ax in facet.axes.flat:
    ax.set_ylim(0.0, 2.0)
    ax.fill_between(
        ax.lines[0].get_xdata(),
        0.0,
        1.0,
        color="green",
        alpha=0.1,
    )
    ax.fill_between(
        ax.lines[0].get_xdata(),
        1.0,
        2.0,
        color="red",
        alpha=0.1,
    )
    ax.set_xlim(1, 50)
    ax.grid()

facet.figure.subplots_adjust(top=0.75)

sns.move_legend(facet, "upper center", ncol=6)
```
